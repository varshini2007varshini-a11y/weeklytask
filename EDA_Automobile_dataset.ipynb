```python
car_counts = (
    df.groupby("make")["price"]
    .count()
    .reset_index()
)

car_counts
```

```python
print(type(plt))
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data=car_counts,
    x="make",
    y="price",
    marker="o"
)

plt.title("Car Count by Make")
plt.xlabel("Car Make")
plt.ylabel("Number of Cars")
plt.xticks(rotation=90)

plt.show()
```
