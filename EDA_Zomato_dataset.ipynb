```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
os.getcwd()
```

```python
dataframe = pd.read_csv(r"C:\\Users\\acer\\Desktop\\weekly task ds\\zomato-data-.csv")
# give your file path
print(dataframe.head())
```

```python
# convert the rate coloum to a float by removing dominator characters.
def handleRate(value):
    value=str(value).split('/')
    value=value[0];
    return float(value)

dataframe['rate']=dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
# getting summary of the dataframe use df.info()
dataframe.info()
```

```python
# checking for missing or null values to identify any data gaps
print(dataframe.isnull().sum())
```

```python
# listed_in (type) coloumn to identify popular restaurant categories
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of restaurant")
```

```python
# votes by restaurant type
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes':grouped_data})
plt.plot(result,c='green',marker='o')
plt.xlabel('Type of restaurant')
plt.ylabel('votes')
```

```python
pivot_table = dataframe.pivot_table(
    index='listed_in(type)',
    columns='online_order',
    aggfunc='size',
    fill_value=0
)

sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='d')

plt.title('Heatmap')
plt.xlabel('online_order')
plt.ylabel('Listed in (Type)')

plt.show()          
```
