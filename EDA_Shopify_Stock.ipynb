```python
df["date"] = pd.to_datetime(df["date"])

print(df["date"])
```

```python
df["price_change"] = df["close"].diff()

print(df["price_change"])
```

```python
df["daily_returns"] = df["close"].pct_change()

print(df["daily_returns"])
```

```python
print(df["volume"])
```

```python
df["upper_bound"] = df["close"].rolling(20).mean() + 2 * df["close"].rolling(20).std()

df["lower_bound"] = df["close"].rolling(20).mean() - 2 * df["close"].rolling(20).std()

print(df[["upper_bound", "lower_bound"]])
```

```python
df[["open", "high", "low", "close"]].plot()

plt.show()
```

```python
df["MA_20"] = df["close"].rolling(20).mean()

df["MA_50"] = df["close"].rolling(50).mean()

print(df[["MA_20", "MA_50"]])
```

```python
df["volatility"] = df["daily_returns"].rolling(20).std()

print(df["volatility"])
```

```python
sns.heatmap(df[["open", "high", "low", "close", "volume"]].corr(), annot=True)

plt.show()
```
