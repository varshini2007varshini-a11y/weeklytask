```python
df.groupby(df["date"].dt.year)["price"].mean()
```

```python
monthly_sales = (
    df.groupby("Sale Month")["id"]
    .count()
    .reset_index()
)

monthly_sales
```

```python
print(type(plt))
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data=monthly_sales,
    x="Sale Month",
    y="id",
    marker="o"
)

plt.title("Monthly House Sales")
plt.xlabel("Sale Month")
plt.ylabel("Number of Houses Sold")

plt.show()
```
