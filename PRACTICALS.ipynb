```python
# Function without arguments and without return value
def hello():
    print("Hello Students")

hello()

# Function with arguments
def add(a, b):
    print("Sum =", a + b)

add(10, 20)

# Function without arguments but with return value
def get_num():
    return 50

print("Returned Value =", get_num())

# Function with arguments and return value
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 4))

# Function with default argument
def greet(name="Student"):
    print("Welcome", name)

greet()
greet("Varshini")

# Function with keyword arguments
def details(name, age):
    print("Name =", name)
    print("Age =", age)

details(age=18, name="Anu")

# Variable length arguments
def marks(*m):
    print("Total Marks =", sum(m))

marks(80, 85, 90)

# Lambda function
square = lambda x: x * x
print("Square =", square(5))

# Recursive function
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print("Factorial =", factorial(5))

# Nested function
def outer():
    print("Outer Function")

    def inner():
        print("Inner Function")

    inner()

outer()
```

```python
# Create a list
numbers = [1, 2, 3, 4, 5]

# Display original list
print("Original List:", numbers)

# First and last elements
print("First Element:", numbers[0])
print("Last Element:", numbers[-1])

# Append an element
numbers.append(6)
print("After Append:", numbers)

# Insert an element
numbers.insert(2, 10)
print("After Insert:", numbers)

# Remove an element
numbers.remove(3)
print("After Remove:", numbers)

# Length of list
print("Length of List:", len(numbers))

# Traverse the list
print("Elements in the List:")
for i in numbers:
    print(i)

# Squares using list comprehension
squares = [x * x for x in numbers]

# Even numbers
even_numbers = [x for x in numbers if x % 2 == 0]

# Odd numbers
odd_numbers = [x for x in numbers if x % 2 != 0]

# Double each element
double_numbers = [x * 2 for x in numbers]

# Nested list (matrix)
matrix = [[1, 2], [3, 4], [5, 6]]

# Flatten the matrix
flat_list = [item for row in matrix for item in row]

# Display generated lists
print("Squares:", squares)
print("Even Numbers:", even_numbers)
print("Odd Numbers:", odd_numbers)
print("Doubled Numbers:", double_numbers)
print("Matrix:", matrix)
print("Flattened List:", flat_list)
```

```python
# Grouping and Aggregating Data using Pandas

import pandas as pd

# Creating DataFrame
data = {
    "Department": ["HR","IT","HR","IT","Sales","Sales"],
    "Employee": ["Amit","Rahul","Priya","Ankit","Sakshi","Riya"],
    "Salary": [30000,50000,35000,60000,45000,40000],
    "Experience": [2,5,3,6,4,2]

}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Grouping by Department
grouped = df.groupby("Department")

# Sum Aggregation
print("\nSum of salary by Department:")
print(grouped["Salary"].sum())

# Mean Aggregation
print("\nAverage Salary by Department:")
print(grouped["Salary"].mean())

# Maximum Salary
print("\nMaximum Salary by Department:")
print(grouped["Salary"].max())

# Minimum Salary
print("\nMinimum Salary by Department:")
print(grouped["Salary"].min())

# Count of Employees
print("\nEmployee Count by Department:")
print(grouped["Employee"].count())

# Multiple Aggregations
print("\nMultiple Aggregations:")
print(
    grouped["Salary"].agg(
        ["sum","mean","max","min"]
    )
)

# Grouping Multiple Columns
print("\nGrouping by Department and Experience:")

group_multi = df.groupby(
    ["Department","Experience"]
)["Salary"].sum()

print(group_multi)

# Iterating Groups
print("\nDisplaying Each Group:")

for name,group in grouped:
    print("\nDepartment:",name)
    print(group)

# Sorting Values
print("\nSorted DataFrame by Salary:")
sorted_df = df.sort_values(by="Salary")

print(sorted_df)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [80, np.nan, 90, np.nan],
    "Age": [20, 21, np.nan, 22]
}

df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame:")
print(df)

# Check missing values
print("\nMissing values:")
print(df.isnull())

# Count missing values
print("\nMissing value count:")
print(df.isnull().sum())

# Fill missing values with default value
df1 = df.fillna(0)
print("\nAfter filling with 0:")
print(df1)

# Fill Marks missing values with mean
df2 = df.copy()
df2["Marks"] = df2["Marks"].fillna(df2["Marks"].mean())
print("\nMarks filled with mean:")
print(df2)

# Forward fill
df3 = df.ffill()
print("\nAfter forward fill:")
print(df3)

# Backward fill
df4 = df.bfill()
print("\nAfter backward fill:")
print(df4)

# Remove rows containing missing values
df5 = df.dropna()
print("\nAfter removing missing rows:")
print(df5)

# Replace a specified value
df6 = df.replace(80, 85)
print("\nAfter replacing 80 with 85:")
print(df6)

# Display DataFrame information
print("\nDataFrame information:")
df.info()
```

```python
# Merge and Join DataFrame using Pandas

import pandas as pd

#----------------CONCATENATE OPERATIONS----------------------(all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID":[101,102,103,105],
    "Marks":[95,88,90,85]
}, index=[0,1,2,3])

print("\nDataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID":[201,202,203,205],
    "Marks":[85,78,80,65]
}, index=[4,5,6,7])

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)


#-------------------MERGE OPERATIONS-------------------(at least one column should be same, others can be different)

#Creating First DataFrame
df1 = pd.DataFrame({
    "ID":[101,102,103,104],
    "Name":["Sakshi","Rahul","Ankit","Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID":[101,102,103,105],
    "Marks":[95,88,90,85]
})
print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\Users\\acer\\Desktop\\weekly task ds\\students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\Users\\acer\\Desktop\\weekly task ds\\students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display the dataset
print("========== DATASET ==========")
print(df)

# 3. Select Marks column
marks = df["Marks"]

# 4. Remove missing values
marks = marks.dropna()

# 5. Count number of observations
count = marks.count()

# 6. Calculate total marks
total = marks.sum()

# 7. Calculate Mean
mean = marks.mean()

# 8. Calculate Median
median = marks.median()

# 9. Calculate Mode
mode = marks.mode()

# 10. Calculate Standard Deviation
std = marks.std()

# 11. Calculate Variance
variance = marks.var()

# 12. Find Minimum
minimum = marks.min()

# 13. Find Maximum
maximum = marks.max()

# 14. Calculate Range
data_range = maximum - minimum

# 15. Calculate First Quartile
q1 = marks.quantile(0.25)

# 16. Calculate Third Quartile
q3 = marks.quantile(0.75)

# 17. Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```
